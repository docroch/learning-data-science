```python
import numpy as np
import pandas as pd
from sklearn import preprocessing
import os
```

```python
dataset = os.path.join('datasets', 'c_data.csv')
df = pd.read_csv(dataset, header=1)
df
```

```python
df.rename(
    columns={
        "CountryName": "Name",
        "CountryCode": "Code",
        "Population growth": "Growth",
        "Total population": "Population",
        "Area (sq. km)": "Area",
    },
    inplace=True,
)
df
```

```python
df.drop(['1', 'Code'], axis=1, inplace=True)
df
```

```python
df.set_index('Name', inplace=True)
df
```

```python
df.replace('?', np.nan, inplace=True)
df
```

```python
df.dtypes
```

```python
df
```

```python
df['Area'] = [float(x) for x in df['Area']]
df['Growth'] = [float(x) for x in df['Growth']]
df['Population'] = [float(x) for x in df['Population']]
df
```

```python
max_pop = df['Population'].max()
df['Population'][df['Population'] == max_pop]
```

```python
df.drop('World', axis=0, inplace=True)
df
```

```python
df.dropna()
```

```python
df.fillna(0)
```

```python
df.fillna({'Growth': 0, 'Population': df.Population.mean(), 'Area': df.Population.mean()})
```

```python
df.ffill()
```

```python
df.bfill()
```

```python
from sklearn.impute import SimpleImputer
```

```python
i = SimpleImputer(missing_values=np.nan, strategy='mean')
i.fit(df)
new_df = i.transform(df)
new_df
```

```python
df[['Growth', 'Population', 'Area']] = new_df
df
```
