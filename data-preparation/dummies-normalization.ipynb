```python
import numpy as np
import pandas as pd
from sklearn import preprocessing
import os
```

```python
dataset = os.path.join('datasets', 'smartphones.csv')
df = pd.read_csv(dataset)
df
```

```python
df.set_index('Name', inplace=True)
df
```

```python
df.drop('Company', axis=1, inplace=True)
df
```

```python
df_dummies = pd.get_dummies(df)
df_dummies
```

```python
from sklearn.preprocessing import scale
```

```python
scale_data = scale(df_dummies)
scale_data
```

```python
df_scaled = pd.DataFrame(scale_data, index=df_dummies.index, columns=df_dummies.columns)
df_scaled
```

```python
from sklearn.preprocessing import normalize
```

```python
norm_data = normalize(df_dummies, norm='l2', axis=0)
norm_data
```

```python
df_norm = pd.DataFrame(norm_data, index=df_dummies.index, columns=df_dummies.columns)
df_norm
```

```python
norm_data = normalize(df_dummies, norm='l1', axis=0)
norm_data
```

```python
df_norm = pd.DataFrame(norm_data, index=df_dummies.index, columns=df_dummies.columns)
df_norm
```

```python
from sklearn.preprocessing import minmax_scale
```

```python
minmax_data = minmax_scale(df_dummies, feature_range=(0, 1))
minmax_data
```

```python
df_minmax = pd.DataFrame(minmax_data, index=df_dummies.index, columns=df_dummies.columns)
df_minmax
```
