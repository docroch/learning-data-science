```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import Lasso
from sklearn.linear_model import Ridge
```

```python
x = np.arange(1, 10)
y = np.array([28, 25, 26, 31, 32, 29, 30, 35, 36])
```

```python
plt.scatter(x, y)
plt.show()
```

```python
y_predict = reg.predict(x)
y_predict
```

```python
dataset = os.path.join('datasets', 'boston.csv')
df = pd.read_csv(dataset)
df
```

```python
df.describe()
```

```python
x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
x
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
reg = LinearRegression()
reg.fit(x_train, y_train)
y_pred = reg.predict(x_test)
```

```python
plt.scatter(y_test, y_pred)
plt.xlabel('Price')
plt.ylabel('Predicted Price')
plt.show()
```

```python
mse = mean_squared_error(y_test, y_pred)
mse
```

```python
new_x = df.iloc[:, [1, 2]].values
new_y = df.iloc[:, -1].values
```

```python
new_x_train, new_x_test, new_y_train, new_y_test = train_test_split(new_x, new_y, test_size=0.3, random_state=42)
```

```python
new_reg = LinearRegression()
new_reg.fit(new_x_train, new_y_train)
new_y_pred = new_reg.predict(new_x_test)
```

```python
mse = mean_squared_error(new_y_test, new_y_pred)
mse
```

```python
plt.scatter(new_y_test, new_y_pred)
plt.xlabel('Price')
plt.ylabel('Predicted Price')
plt.show()
```

```python
reg = LinearRegression()
cv_score = cross_val_score(reg, x, y, cv=5)
cv_score
```

```python
np.mean(cv_score)
```

```python
lasso = Lasso(alpha=0.1)
lasso.fit(x, y)
lasso_coef = lasso.coef_
lasso_coef
```

```python
plt.plot(range(len(lasso_coef)), lasso_coef)
plt.xticks(range(len(lasso_coef)), df.columns[:-1])
plt.ylabel('Coefficents')
plt.grid()
plt.show()
```

```python
lasso = Lasso(alpha=0.1)
lasso.fit(x_train, y_train)
y_lasso = lasso.predict(x_test)
```

```python
lasso_mse = mean_squared_error(y_test, y_lasso)
lasso_mse
```

```python
ridge = Ridge(alpha=0.1)
ridge.fit(x_train, y_train)
y_ridge = ridge.predict(x_test)
```

```python
ridge_mse = mean_squared_error(y_test, y_ridge)
ridge_mse
```
