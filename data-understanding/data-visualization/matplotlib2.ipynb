```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
years =  [1960, 1970, 1980, 1990, 2000, 2010, 2020]
iranPop = [21.19, 28.51, 38.67, 56.23, 66.13, 74.75, 80.29]
```

```python
plt.figure(figsize=(8, 6), dpi=100)
plt.plot(years, iranPop)
plt.xlabel('Years')
plt.ylabel('Population')
plt.yticks(iranPop, [f'{round(x)}m' for x in iranPop])
plt.show()
```

```python
plt.scatter(years, iranPop)
plt.show()
```

```python
cityNames = ['Tehran', 'Mashhad', 'Isfahan', 'Karaj', 'Tabriz', 'Shiraz']
cityPop = [7153309, 2307177, 1547164, 1448075, 1424641, 1249942]
```

```python
plt.hist(cityPop)
plt.show()
```

```python
plt.pie(cityPop, labels=cityNames)
plt.show()
```

```python
popsize = np.array(cityPop)/10000
colors = ['pink', 'blue', 'red', 'orange', 'green', 'yellow']
plt.scatter(cityNames, cityPop, s=popsize, c=colors)
plt.margins(0.1)
plt.title('Iran Population')
plt.yticks([1000000, 2000000, 3000000, 7000000], ['1m', '2m', '3m', '7m'])
plt.show()
```

```python
turkeyPop = [27.47, 34.88, 43.98, 53.92, 63.24, 72.33, 79.51]
```

```python
plt.plot(years ,iranPop, ls='-', marker='+', mew=8)
plt.plot(years ,turkeyPop, ls='--', marker='+', mew=8)
plt.title('Iran vs Turkey Population')
plt.legend(['Iran', 'Turkey'])
plt.xlabel('Year')
plt.ylabel('Population')
plt.yticks(range(20, 90, 10), [f'{round(x)}m' for x in range(20, 90, 10)])
plt.grid()
plt.annotate('Iran in 1990', xytext=(1995, 40), xy=(1990, 56.23), arrowprops={'facecolor': 'silver', 'width': 3}, fontsize=14)
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.plot(years ,iranPop, ls='-', marker='+', mew=8, c='blue')
plt.title('Iran')
plt.subplot(1, 2, 2)
plt.plot(years ,turkeyPop, ls='--', marker='+', mew=8, c='orange')
plt.title('Turkey')

plt.show()
```
