```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
dataset = os.path.join('datasets', 'smartphones.csv')
df = pd.read_csv(dataset)
df
```

```python
def ECDF(data):
    n = len(data)
    x = np.sort(data)
    y = np.arange(1, n+1)/n
    return x, y

x, y = ECDF(df.inch)
```

```python
x
```

```python
y
```

```python
plt.figure(figsize=(9, 6))
plt.scatter(x, y)
plt.xticks(x)
plt.yticks([round(i, 2) for i in y])
plt.xlabel('Inch')
plt.ylabel('ECDF')
plt.grid()
plt.show()
```
