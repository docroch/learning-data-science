```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets
from sklearn.svm import SVC
from sklearn import metrics
```

```python
data = datasets.load_digits()
```

```python
plt.subplot()
plt.imshow(data.images[0], cmap=plt.cm.gray_r)
plt.show()
```

```python
data.target[0]
```

```python
images_and_labels = list(zip(data.images, data.target))
images_and_labels[0]
```

```python
for i, (image, label) in enumerate(images_and_labels[:4]):
    plt.subplot(2, 4, i + 1)
    plt.imshow(image, cmap=plt.cm.grey_r)
    plt.title(label=label)
```

```python
x = data.images.reshape((len(data.images), -1))
x
```

```python
y = data.target
y
```

```python
svm_classifier = SVC(random_state=3432, C=0.5)
```

```python
svm_classifier.fit(x, y)
```
