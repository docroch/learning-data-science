```python
fig = px.box(
    df,
    x='fuelCost08',
    y='class_summary',
    color='class_summary',
    labels={'fuelCost08': 'fuel cost'}
)
fig.update_layout(showlegend=False)
fig.show()
```

```python
fig = px.strip(
    df,
    x='fuelCost08',
    y='class_summary',
    color='class_summary',
    labels={'fuelCost08': 'fuel cost'}
)
fig.update_layout(showlegend=False)
fig.show()
```

```python
avg_cost = round(df['fuelCost08'].mean(),2)
fig = px.histogram(
    df,
    x='fuelCost08',
    color='class_summary',
    nbins=40,
    title='Fuel Cost Distribution',
    color_discrete_sequence=px.colors.qualitative.G10
)
fig.update_yaxes(title_text='Number of Cars')
fig.add_vline(x=avg_cost, annotation_text=f'Average ${avg_cost}')
fig.show()
```

```python
fig = px.histogram(
    df,
    x='fuelCost08',
    color='class_summary',
    nbins=40,
    title='Fuel Cost Distribution',
    marginal='box',
    color_discrete_sequence=px.colors.qualitative.Bold
)
fig.show()
```

```python
fig = px.scatter(
    df.query('co2 > 0'),
    x='fuelCost08',
    y='co2',
    hover_name='model',
    size='cylinders',
    color='cylinders',
    hover_data=['make', 'year', 'cylinders']
)
fig.show()
```

```python
fig = px.scatter(
    df.query('co2 > 0'),
    x='fuelCost08',
    y='co2',
    hover_name='model',
    size='cylinders',
    color='cylinders',
    hover_data={'make': True, 'year': True, 'cylinders': ':.0f', 'fuelCost08': ':$.0f'}
)
fig.show()
```

```python
fig = px.scatter_matrix(
    df,
    dimensions=['cylinders', 'fuelCost08'],
    color='make'
)
fig.show()
```

```python
avg_by_year = df.groupby(['year'], as_index=False).agg({'fuelCost08': 'mean'}).round(2)
avg_by_year
```

```python
fig = px.scatter(avg_by_year, x='year', y='fuelCost08', trendline='ols')
fig.show()
```

```python
fig = px.scatter(avg_by_year, x='year', y='fuelCost08', trendline='lowess')
fig.show()
```

```python
fig = px.sunburst(
    df.query('year > 2015'),
    path=['year', 'class_summary']
)
fig.show()
```

```python
fig = px.density_heatmap(df, x='year', y='class_summary')
fig.show()
```

```python
fig = px.density_heatmap(
    df,
    x='year', 
    y='highway08',
    marginal_y='histogram',
    template='plotly_white',
    range_y=[10, 40]
)
fig.show()
```

```python
fig = px.box(
    df,
    x='date_range',
    y='fuelCost08',
    facet_col='fuel_type_summary',
    facet_col_wrap=2,
    boxmode='group',
    color='drive',
    points=False,
    height=700
)
fig.show()
```

```python
fig = px.histogram(
    df,
    x='fuelCost08',
    facet_col='class_summary',
    facet_col_wrap=4,
    nbins=40
)
fig.for_each_annotation(lambda a: a.update(text=a.text.split('=')[-1]))
fig.show()
```

```python
fig = px.strip(
    df,
    x='fuelCost08',
    y='class_summary',
    color='class_summary',
    range_x=[300, 4000],
)
fig.update_layout(showlegend=False)
fig.update_yaxes(title='Vehicle Class')
fig.update_xaxes(title='Estimated Fuel Cost')
fig.add_annotation(
    x=1400,
    y='Van',
    text='Outlier Value',
    showarrow=True,
    arrowhead=1
)
fig.add_vrect(
    x0=3500,
    x1=4000,
    fillcolor='LightSalmon',
    opacity=0.5,
    layer='below',
    line_width=0,
    annotation_text='Inefficient Vehicles',
    annotation_position='bottom'
)

fig.show()
```
